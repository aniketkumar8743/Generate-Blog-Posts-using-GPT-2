```python
# what this line of code do GPT2Tokenizer-> take input from user and convert into number , GPT2LMHeadModel-> when number collected then it passed 
#through this Modeland it will generate words
from transformers import GPT2LMHeadModel, GPT2Tokenizer
```

## 1. Load Model

```python
tokenizer.eos_token_id
```

```python
tokenizer = GPT2Tokenizer.from_pretrained('gpt2-large')
model = GPT2LMHeadModel.from_pretrained("gpt2-large", pad_token_id = tokenizer.eos_token_id)
```

```python
tokenizer.decode(tokenizer.eos_token_id)
```

## 2. Tokenize Sentences

```python
sentence = "Aritificial intelligence lacks the ability to draw from personal experiences"
input_ids = tokenizer.encode(sentence,return_tensors='pt')
```

```python
input_ids
```

```python
input_ids[0][2]
```

```python
tokenizer.decode(input_ids[0][2])
```

## 3.Generate and Decode Text

num_beams -> used to find from a paticually word to find out next most similary or fit  words,
no_repeat_ngram_size -> it help a model to repeat the again and again repeat the word
early_stopping -> it stop when we not getting a great output so it stop

```python
output = model.generate(
    input_ids,
    max_length=1000, 
    
    num_beams=5,
    no_repeat_ngram_size=2,
    early_stopping=False,  
    do_sample=True,  
    temperature=0.9,  
    top_p=0.95  
)
```

```python
tokenizer.decode(output[0], skip_special_tokens = True)
```

```python
output
```

## 4. Output Result

```python
text = tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
text = text.replace("\n", "")
```
